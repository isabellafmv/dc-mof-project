```python
import re
import json
import pickle
from pathlib import Path

import numpy as np
import pandas as pd
from tqdm.auto import tqdm
from mofdb_client import fetch
```

```python
# ---------------------------------------------------------------------------
# Helper functions
# ---------------------------------------------------------------------------

# Pressure points present in hMOF GCMC data (bar)
CO2_PRESSURES = [0.01, 0.05, 0.1, 0.5, 2.5]
PRESSURE_TOL  = 1e-3   # absolute tolerance for pressure matching (bar)

def _col(p: float) -> str:
    """Column name for a CO2 pressure point, e.g. 'co2_mol_kg_0.05bar'."""
    return f"co2_mol_kg_{p}bar"


def _parse_topology(mofid: str | None) -> str | None:
    """
    Extract net topology from a MOF-ID string.
    MOF-ID format: '<smiles_fragment(s)> MOFid-v1.<topology>.<cat>'
    """
    if not mofid:
        return None
    m = re.search(r"MOFid-v1\.([^.\s]+)", mofid)
    return m.group(1) if m else None


def _parse_metal_node(mofkey: str | None) -> str | None:
    """
    Extract metal-node label from a MOF-key string.
    MOF-key format: '<Metal>.<InChIKey1>.<InChIKey2>...MOFkey-v1.<topology>'
    The first token before the first '.' is the metal symbol(s).
    """
    if not mofkey:
        return None
    return mofkey.split(".")[0]


def _extract_co2_uptake(mof) -> dict:
    """
    Return a dict {col_name: uptake_mol_kg} for all CO2 pressure points.
    Picks the 298 K pure-CO2 simulation isotherm; falls back to any CO2 isotherm.
    If multiple isotherms exist, prefers simulated (simin='True') ones.
    """
    uptake = {_col(p): np.nan for p in CO2_PRESSURES}

    # Filter to CO2-only isotherms at 298 K
    co2_isos = [
        iso for iso in mof.isotherms
        if len(iso.adsorbates) == 1
        and iso.adsorbates[0].name == "CarbonDioxide"
        and abs(iso.temperature - 298.0) < 1.0
        and iso.pressureUnits.lower() in ("bar",)
        and "mol/kg" in iso.adsorptionUnits.lower()
    ]

    # Prefer simulated isotherms
    sim_isos = [iso for iso in co2_isos if str(iso.simin).lower() == "true"]
    chosen = sim_isos[0] if sim_isos else (co2_isos[0] if co2_isos else None)

    if chosen is None:
        return uptake

    for pt in chosen.isotherm_data:
        for target_p in CO2_PRESSURES:
            if abs(pt.pressure - target_p) < PRESSURE_TOL:
                # total_adsorption is the sum; for pure-CO2 it equals species_data[0].adsorption
                uptake[_col(target_p)] = pt.total_adsorption
                break

    return uptake


def _mof_to_row(mof) -> dict:
    """Convert a Mof object into a flat dict suitable for a DataFrame row."""
    sa_m2g   = mof.surface_area_m2g
    sa_m2cm3 = mof.surface_area_m2cm3

    # Density: rho = SA[m²/cm³] / SA[m²/g]  (algebraically: g/cm³)
    if sa_m2g and sa_m2cm3 and sa_m2g > 0:
        density = sa_m2cm3 / sa_m2g
    else:
        density = np.nan

    row = {
        "name"               : mof.name,
        "mofdb_id"           : mof.id,
        "database"           : mof.database,
        "pld"                : mof.pld,
        "lcd"                : mof.lcd,
        "surface_area_m2g"   : sa_m2g,
        "surface_area_m2cm3" : sa_m2cm3,
        "void_fraction"      : mof.void_fraction,
        "density_g_cm3"      : density,
        "topology"           : _parse_topology(mof.mofid),
        "metal_node"         : _parse_metal_node(mof.mofkey),
        "elements"           : ",".join(str(e) for e in mof.elements),
        "mofid"              : mof.mofid,
        "mofkey"             : mof.mofkey,
    }
    row.update(_extract_co2_uptake(mof))
    return row
```

```python
# ---------------------------------------------------------------------------
# Download with checkpointing
# ---------------------------------------------------------------------------

CHECKPOINT_PATH = Path("hmof_checkpoint.pkl")
PARQUET_PATH    = Path("hmof_dataset.parquet")
CHECKPOINT_EVERY = 1000
HMOF_TOTAL = 137953  # known total; download is skipped if checkpoint is already complete

def download_hmof(resume: bool = True) -> list[dict]:
    rows: list[dict] = []

    if resume and CHECKPOINT_PATH.exists():
        with open(CHECKPOINT_PATH, "rb") as f:
            rows = pickle.load(f)
        print(f"Loaded {len(rows):,} structures from checkpoint.")
        if len(rows) >= HMOF_TOTAL:
            print("Download already complete, skipping API fetch.")
            return rows

    start_idx = len(rows)
    with tqdm(desc="Downloading hMOF", unit=" MOF", initial=start_idx,
              miniters=100, dynamic_ncols=True) as pbar:
        for mof in fetch(database="hMOF"):
            rows.append(_mof_to_row(mof))
            pbar.update(1)

            if len(rows) % CHECKPOINT_EVERY == 0:
                with open(CHECKPOINT_PATH, "wb") as f:
                    pickle.dump(rows, f)

    with open(CHECKPOINT_PATH, "wb") as f:
        pickle.dump(rows, f)

    print(f"\nDownload complete. Total structures: {len(rows):,}")
    return rows


```

```python
rows = download_hmof(resume=True)
```

```python
# ---------------------------------------------------------------------------
# Assemble DataFrame and save (all rows, no dropping)
# ---------------------------------------------------------------------------

df = pd.DataFrame(rows)

# Enforce sensible dtypes
float_cols = ["pld", "lcd", "surface_area_m2g", "surface_area_m2cm3",
              "void_fraction", "density_g_cm3"] + [_col(p) for p in CO2_PRESSURES]
for c in float_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")

df["mofdb_id"] = pd.to_numeric(df["mofdb_id"], errors="coerce").astype("int64")

# Reorder columns
geo_cols = ["pld", "lcd", "surface_area_m2g", "surface_area_m2cm3",
            "void_fraction", "density_g_cm3"]
co2_cols = [_col(p) for p in CO2_PRESSURES]
meta_cols = ["topology", "metal_node", "elements", "mofid", "mofkey", "database", "mofdb_id"]

df = df[["name"] + geo_cols + co2_cols + meta_cols]

print(f"Assembled: {df.shape[0]:,} rows × {df.shape[1]} columns")
df.head(3)
```

```python
# Missing value analysis

total = len(df)

missing = pd.DataFrame({
    "missing_n"  : df.isnull().sum(),
    "missing_pct": (df.isnull().mean() * 100).round(2),
}).sort_values("missing_pct", ascending=False)

missing = missing[missing["missing_n"] > 0]   # only show columns that have gaps

print(f"Total structures: {total:,}\n")
print(missing.to_string())
```

```python
# ---------------------------------------------------------------------------
# Drop rows with missing CO2 targets or topology, then save to Parquet
# ---------------------------------------------------------------------------

# Replace sentinel strings with NaN so they get caught by dropna
TOPOLOGY_SENTINELS = {"NA", "ERROR", "UNKNOWN"}
df["topology"] = df["topology"].where(~df["topology"].isin(TOPOLOGY_SENTINELS), other=None)

drop_cols = co2_cols + ["topology"]
df = df.dropna(subset=drop_cols)
print(f"After dropping CO2/topology NAs: {len(df):,} rows (removed {total - len(df):,})")
remaining = sorted(df["topology"].unique())
print(f"Remaining topologies: {remaining}")

# Save via subprocess to avoid pyarrow kernel-conflict bug
import subprocess, sys
CSV_PATH = Path("hmof_dataset.csv")
df.to_csv(CSV_PATH, index=False)

PARQUET_CLEANED_PATH = Path("hmof_dataset_cleaned.parquet")

result = subprocess.run(
    [sys.executable, "-c",
     f"import pandas as pd; pd.read_csv('{CSV_PATH}').to_parquet('{PARQUET_CLEANED_PATH}', index=False)"],
    capture_output=True, text=True
)
if result.returncode != 0:
    print("Parquet conversion failed:", result.stderr)
else:
    CSV_PATH.unlink()
    print(f"Saved → {PARQUET_CLEANED_PATH}")

```

```python
# Quick sanity checks

print("Shape")
print(df.shape)

print("\nDtypes")
print(df.dtypes)

print("\nMissing values (%)")
print((df.isnull().mean() * 100).round(2).to_string())

print("\nDescriptor statistics")
print(df[geo_cols + co2_cols].describe().round(3).to_string())

print(f"\nUnique topologies ({df['topology'].nunique()})")
print(df["topology"].value_counts().head(20).to_string())

print(f"\nUnique metal nodes ({df['metal_node'].nunique()})")
print(df["metal_node"].value_counts().head(20).to_string())
```

```python
# check for data distribution

import matplotlib.pyplot as plt
from scipy.stats import skew

plot_cols = ["pld", "lcd", "surface_area_m2g", "surface_area_m2cm3",
             "void_fraction", "density_g_cm3"] + co2_cols

n_cols = 3
n_rows = -(-len(plot_cols) // n_cols)  # ceiling division

fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, n_rows * 3.5))
axes = axes.flatten()

for ax, col in zip(axes, plot_cols):
    data = df[col].dropna()
    sk = skew(data)
    ax.hist(data, bins=80, color="steelblue", edgecolor="none", alpha=0.8)
    ax.set_title(f"{col}\nskewness = {sk:.2f}", fontsize=9)
    ax.set_xlabel(col, fontsize=8)
    ax.set_ylabel("count", fontsize=8)
    ax.tick_params(labelsize=7)

# hide unused axes
for ax in axes[len(plot_cols):]:
    ax.set_visible(False)

plt.suptitle("Distribution of geometric descriptors & CO₂ uptake", fontsize=12, y=1.01)
plt.tight_layout()
plt.show()
```

```python
# log1p transform for skewed columns

log_cols = ["pld", "lcd", "density_g_cm3"] + co2_cols

for col in log_cols:
    df[f"log1p_{col}"] = np.log1p(df[col])

log_transformed_cols = [f"log1p_{c}" for c in log_cols]

# Compare distributions before/after
fig, axes = plt.subplots(len(log_cols), 2, figsize=(12, len(log_cols) * 2.8))

for i, col in enumerate(log_cols):
    data_raw = df[col].dropna()
    data_log = df[f"log1p_{col}"].dropna()

    sk_raw = skew(data_raw)
    sk_log = skew(data_log)

    axes[i, 0].hist(data_raw, bins=80, color="steelblue", edgecolor="none", alpha=0.8)
    axes[i, 0].set_title(f"{col}  (skew={sk_raw:.2f})", fontsize=9)
    axes[i, 0].set_ylabel("count", fontsize=8)
    axes[i, 0].tick_params(labelsize=7)

    axes[i, 1].hist(data_log, bins=80, color="darkorange", edgecolor="none", alpha=0.8)
    axes[i, 1].set_title(f"log1p({col})  (skew={sk_log:.2f})", fontsize=9)
    axes[i, 1].tick_params(labelsize=7)

plt.suptitle("Before vs after log1p transform", fontsize=12, y=1.01)
plt.tight_layout()
plt.show()

print("log1p columns added:", log_transformed_cols)
```

```python
import matplotlib.pyplot as plt
from scipy.stats import entropy

# ---------------------------------------------------------------------------
# Do we need stratification? Explore class imbalance in topology & metal_node
# ---------------------------------------------------------------------------

fig, axes = plt.subplots(1, 2, figsize=(14, 4))

for ax, col in zip(axes, ['topology', 'metal_node']):
    counts = df[col].value_counts().sort_values(ascending=False)
    pct    = counts / counts.sum() * 100

    ax.bar(range(len(counts)), pct.values, color='steelblue', edgecolor='none')
    ax.set_xticks(range(len(counts)))
    ax.set_xticklabels(counts.index, rotation=45, ha='right', fontsize=8)
    ax.set_ylabel('% of dataset')
    ax.set_title(f'{col} distribution')

    norm_entropy = entropy(counts.values, base=2) / (len(counts) if len(counts) > 1 else 1)

    print(f'\n{col}')
    print(f'  classes:            {len(counts)}')
    print(f'  largest class:      {counts.index[0]}  ({pct.iloc[0]:.1f}%)')
    print(f'  smallest class:     {counts.index[-1]}  ({pct.iloc[-1]:.2f}%,  n={counts.iloc[-1]})')
    print(f'  normalized entropy: {norm_entropy:.3f}  (1=uniform, 0=one class dominates)')
    print(f'  classes with n<10:  {(counts < 10).sum()}')
    print(f'  classes with n<2:   {(counts < 2).sum()}  <- too small to stratify')

plt.tight_layout()
plt.show()

# Does topology correlate with CO2 uptake? If yes, stratification matters more.
print('\nMean CO2 uptake @ 2.5 bar per topology (sorted):')
topo_means = df.groupby('topology')['co2_mol_kg_2.5bar'].agg(['mean', 'std', 'count'])
print(topo_means.sort_values('mean', ascending=False).to_string())
```
